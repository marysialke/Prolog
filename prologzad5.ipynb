# Zadanie 5: System rekomendacji książek

Cel zadania: Stwórz system rekomendacji książek, który uwzględnia preferencje użytkowników, gatunki literackie oraz oceny książek. Twój system powinien być w stanie rekomendować książki danego gatunku użytkownikowi na podstawie jego preferencji oraz wcześniej przyznanych ocen.

Struktura bazy danych:

* Książki: Informacje o książkach obejmujące tytuł, autora i gatunek.
* Oceny: Oceny przyznawane przez użytkowników poszczególnym książkom.
* Preferencje: Preferowane gatunki literackie wybranych użytkowników.

Reguły do zdefiniowania:

* polecane_ksiazki(X, Gatunek): Ta reguła powinna identyfikować książki z określonego gatunku, które są wysoko oceniane przez użytkownika X oraz odpowiadają jego preferencjom.

```python
% bóg komenda czyli
:- dynamic ocena/3.

% Mangi: tytuł, autor, gatunek; mangi bo się znam na tym najlepiej
manga('Given', 'Natsuki Kizu', boys_love).
manga('Ten Count', 'Rihito Takarai', yaoi).
manga('Dr. Stone', 'Riichirou Inagaki', przygodowe).
manga('Mr. Villain Has a Day Off', 'Taro Yamaguchi', komedia).
manga('The Guy She Was Interested in Wasnt a Guy at All', 'Kouhei Azano', girls_love).

% Ocenki, tak jestem biased
ocena('user1', 'Dr. Stone', 5).
ocena('user4', 'The Guy She Was Interested in Wasnt a Guy at All', 5).
ocena('user6', 'Mr. Villain Has a Day Off', 5).
ocena('user7', 'Ten Count', 5).
ocena('user5', 'The Guy She Was Interested in Wasnt a Guy at All', 5).

% Preferowane gatunki
preferencja('user2', yaoi).
preferencja('user1', przygodowa).
preferencja('user7', yaoi).
preferencja('user4', komedia).
preferencja('user5', girls_love).

% Co polecam (nie wiem jak użyć bagof itd ale znam liste i że musi mieć 2 rzeczy do siebie więc działa tak)
polecane_mangi(Użytkownik, Tytul) :-
    preferencja(Użytkownik, Gatunek),
    findall(Tytul, (manga(Tytul, _, Gatunek), ocena(Użytkownik, Tytul, Ocena), Ocena >= 3), Wyniki),
    list_to_set(Wyniki, Ocena),
    format('Dla ~w polecamy mangi w gatunku ~w.', [Użytkownik, Gatunek]).

% Ocenki; tu dodajemy
dodaj_ocene(Użytkownik, Tytul, Ocena) :-
    assertz(ocena(Użytkownik, Tytul, Ocena)).

% tu usuwamy
usun_ocene(Użytkownik, Tytul) :-
    retract(ocena(Użytkownik, Tytul, _)).
usun_ocene(_, _). % raczej działa, ale u mnie wszędzie są 5, także nie będe zmieniać na gorsze :)

% chyba działa ogółem, dziwne to troszkę
```
